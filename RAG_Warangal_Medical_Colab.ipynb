## 🧠 Step 2: Initialize the Modular RAG Backend Engine
We import and initialize `WarangalMedicalRAG` from `backend.py`. The backend loads 12,000 clinical records and builds the retrieval index.

```python
from backend import WarangalMedicalRAG, check_red_flags

# Initialize RAG engine (TF-IDF instant retrieval + optional FAISS dense vectors)
rag = WarangalMedicalRAG(enable_faiss=True)
print('\n✅ RAG Backend Engine initialized and ready for queries!')
```

## 🔍 Step 3: Run Interactive Queries in Colab
You can test any medical question or symptom query. Let's run 3 clinical test cases:
1. **Primary Care Query** (Asthma in Parkal)
2. **Specialized & Filtered Query** (Dermatology in Hanamkonda)
3. **Emergency Red-Flag Trigger** (Severe chest pain in Kazipet)

```python
# Test Case 1: Primary Care Query in Parkal
query_1 = "Child in Parkal having severe asthma and wheezing"
result_1 = rag.query(query_1, engine_type="TF-IDF + Cosine Similarity", top_k=3)

print(f"[Query]: {result_1['query']}")
print(f"[Latency]: {result_1['latency_ms']:.2f} ms")
print(f"[Emergency]: {result_1['is_emergency']}")
print("\n--- Grounded Clinical Guidance ---")
print(result_1['response_text'])
```

```python
# Test Case 2: Filtered Query for Dermatology in Hanamkonda
query_2 = "Intense skin itching with dry red patches"
result_2 = rag.query(
    query_2,
    engine_type="TF-IDF + Cosine Similarity",
    dept_filter="Dermatology",
    locality_filter="Hanamkonda",
    top_k=3
)

print(f"[Query]: {result_2['query']}")
print(f"[Latency]: {result_2['latency_ms']:.2f} ms")
print("\n--- Grounded Clinical Guidance ---")
print(result_2['response_text'])
```

```python
# Test Case 3: Emergency Red-Flag Triage in Kazipet
query_3 = "Severe chest pain, heavy sweating, and breathing distress in Kazipet"
result_3 = rag.query(query_3, engine_type="TF-IDF + Cosine Similarity", top_k=3)

print(f"[Query]: {result_3['query']}")
print(f"[Emergency Status]: {'🚨 CRITICAL RED-FLAG ALERT' if result_3['is_emergency'] else 'Normal'}")
print(f"[Matched Triggers]: {result_3['matched_triggers']}")
print("\n--- Grounded Clinical Guidance ---")
print(result_3['response_text'])
```
